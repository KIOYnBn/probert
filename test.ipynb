```python
import keras_tuner as kt
from tensorflow import keras

# 1. 定义模型构建函数（必须包含 hp 参数）
def build_model(hp):
    model = keras.Sequential()
    
    # 调整全连接层神经元数量（32-512之间随机尝试）
    hp_units = hp.Int('units', min_value=32, max_value=512, step=32)
    model.add(keras.layers.Dense(units=hp_units, activation='relu'))
    
    # 调整学习率（从预设值中选择）
    hp_learning_rate = hp.Choice('learning_rate', values=[1e-2, 1e-3, 1e-4])
    
    # 编译模型
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=hp_learning_rate),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# 2. 初始化调优器
tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=10,          # 最大尝试次数
    executions_per_trial=3, # 每个配置训练3次（减少随机性影响）
    directory='my_dir',
    project_name='helloworld'
)

# 3. 加载数据（示例使用MNIST）
(x, y), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train, x_val = x[:50000] / 255.0, x[50000:] / 255.0
y_train, y_val = y[:50000], y[50000:]

# 4. 执行超参数搜索
tuner.search(x_train, y_train,
             epochs=5,
             validation_data=(x_val, y_val))

# 5. 获取最佳模型
best_model = tuner.get_best_models(num_models=1)[0]
best_hyperparameters = tuner.get_best_hyperparameters()[0]

print(f"最佳神经元数: {best_hyperparameters.get('units')}")
print(f"最佳学习率: {best_hyperparameters.get('learning_rate')}")
```
